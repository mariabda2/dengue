```python
# Configuración global del proyecto
import os
from pathlib import Path
import matplotlib.pyplot as plt
import seaborn as sns

PROJECT_ROOT = Path('../../')
DATA_RAW     = PROJECT_ROOT / 'data' / 'raw'
DATA_INTERIM = PROJECT_ROOT / 'data' / 'interim'
DATA_PROC    = PROJECT_ROOT / 'data' / 'processed'
OUT_FIGURES  = PROJECT_ROOT / 'outputs' / 'figures'
OUT_TABLES   = PROJECT_ROOT / 'outputs' / 'tables'
OUT_MODELS   = PROJECT_ROOT / 'outputs' / 'models'
CHECKPOINTS  = PROJECT_ROOT / 'outputs' / 'checkpoints'

for p in [DATA_RAW,DATA_INTERIM,DATA_PROC,OUT_FIGURES,OUT_TABLES,OUT_MODELS,CHECKPOINTS]:
    p.mkdir(parents=True, exist_ok=True)

RANDOM_STATE = 42
TEST_SIZE    = 0.20
CV_FOLDS     = 5

COLORS = {
    'primary':   '#2C7BB6',
    'secondary': '#D7191C',
    'accent':    '#FDAE61',
    'neutral':   '#7F7F7F',
    'con_signos':'#E74C3C',
    'sin_signos':'#3498DB',
    'grave':     '#8E44AD',
}

plt.rcParams.update({
    'figure.dpi':150, 'savefig.dpi':300,
    'figure.facecolor':'white','axes.facecolor':'white',
    'axes.spines.top':False,'axes.spines.right':False,
    'axes.titlesize':14,'axes.titleweight':'bold',
    'axes.labelsize':12,'xtick.labelsize':10,
    'ytick.labelsize':10,'legend.frameon':False,
})
sns.set_theme(style='whitegrid', font_scale=1.05)
print('Configuracion lista')
print(f'  RANDOM_STATE : {RANDOM_STATE}')
print(f'  TEST_SIZE    : {TEST_SIZE}')
print(f'  CV_FOLDS     : {CV_FOLDS}')
```

```python
# Verificar archivo de datos
import pandas as pd
DATA_FILE = DATA_RAW / 'datos_dengue_bucaramanga_depurado.csv'
if DATA_FILE.exists():
    df_check = pd.read_csv(DATA_FILE, encoding='utf-8-sig', nrows=3)
    print(f'Archivo encontrado: {DATA_FILE.name}')
    print(f'  Columnas: {df_check.shape[1]}')
    print(f'  Primeras columnas: {list(df_check.columns[:8])}')
else:
    print('AVISO: Coloca el CSV en data/raw/')
    print(f'  Ruta esperada: {DATA_FILE.resolve()}')
```
