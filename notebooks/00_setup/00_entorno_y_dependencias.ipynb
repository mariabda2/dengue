````python
# ── CONFIGURACION DE RUTA ────────────────────────────────────────────
# Edita PROJECT_ROOT con la ruta real en tu maquina
# ─────────────────────────────────────────────────────────────────────
import os, sys
from pathlib import Path

PROJECT_ROOT = Path(r'D:\Documentos\GitHub\dengue')  # <-- edita aqui

assert PROJECT_ROOT.exists(), (
    f'ERROR: la ruta no existe: {PROJECT_ROOT}\n'
    f'Edita PROJECT_ROOT en esta celda con la ruta correcta'
)
print(f'OK: {PROJECT_ROOT}')

DATA_RAW     = PROJECT_ROOT / 'data' / 'raw'
DATA_INTERIM = PROJECT_ROOT / 'data' / 'interim'
DATA_PROC    = PROJECT_ROOT / 'data' / 'processed'
OUT_FIGURES  = PROJECT_ROOT / 'outputs' / 'figures'
OUT_TABLES   = PROJECT_ROOT / 'outputs' / 'tables'
OUT_MODELS   = PROJECT_ROOT / 'outputs' / 'models'
CHECKPOINTS  = PROJECT_ROOT / 'outputs' / 'checkpoints'
DATA_FILE    = DATA_RAW / 'datos_dengue_bucaramanga_depurado.csv'

for p in [DATA_RAW, DATA_INTERIM, DATA_PROC,
          OUT_FIGURES, OUT_TABLES, OUT_MODELS, CHECKPOINTS]:
    p.mkdir(parents=True, exist_ok=True)

print(f'CSV: {DATA_FILE}')
print(f'CSV existe: {DATA_FILE.exists()}')
print('Directorios creados')
````

````python
# Deteccion automatica de la raiz del proyecto
# Busca la carpeta que contiene data/raw/ y notebooks/ hacia arriba
import os, sys
from pathlib import Path

def encontrar_raiz():
    # En VS Code, __vsc_ipynb_file__ da la ruta del notebook
    try:
        nb_path = Path(globals()['__vsc_ipynb_file__']).resolve()
    except KeyError:
        nb_path = Path(os.getcwd())
    candidato = nb_path.parent
    for _ in range(8):
        if (candidato / 'data' / 'raw').exists():
            return candidato
        candidato = candidato.parent
    return nb_path.parent

PROJECT_ROOT = encontrar_raiz()
DATA_FILE    = PROJECT_ROOT / 'data' / 'raw' / 'datos_dengue_bucaramanga_depurado.csv'

print(f'Raiz detectada : {PROJECT_ROOT}')
print(f'CSV esperado   : {DATA_FILE}')
print()

if DATA_FILE.exists():
    print('OK: archivo CSV encontrado')
else:
    print('='*60)
    print('ACCION REQUERIDA: el CSV no esta en data/raw/')
    print('='*60)
    print()
    print('Copia el archivo aqui:')
    print(f'  {DATA_FILE}')
    print()
    print('Desde el Explorador de Windows: clic derecho -> copiar')
    print(f'Pegar en: {DATA_FILE.parent}')
````

````python
# Crear directorios del proyecto
DATA_INTERIM = PROJECT_ROOT / 'data' / 'interim'
DATA_PROC    = PROJECT_ROOT / 'data' / 'processed'
OUT_FIGURES  = PROJECT_ROOT / 'outputs' / 'figures'
OUT_TABLES   = PROJECT_ROOT / 'outputs' / 'tables'
OUT_MODELS   = PROJECT_ROOT / 'outputs' / 'models'
CHECKPOINTS  = PROJECT_ROOT / 'outputs' / 'checkpoints'

for p in [DATA_INTERIM, DATA_PROC, OUT_FIGURES,
          OUT_TABLES, OUT_MODELS, CHECKPOINTS]:
    p.mkdir(parents=True, exist_ok=True)

print('Directorios creados/verificados:')
for nombre, ruta in [('data/interim', DATA_INTERIM),
    ('outputs/figures', OUT_FIGURES), ('outputs/tables', OUT_TABLES),
    ('outputs/models', OUT_MODELS), ('outputs/checkpoints', CHECKPOINTS)]:
    print(f'  OK  {nombre}')
````
